# Linear Regression Assignment #

```python
# import the libraries:
import pandas as pd
dataset=pd.read_csv('insurance_pre.csv')
dataset.head()
```

```python
dataset.age.dtype
```

```python
dataset.smoker.dtype
```

```python
dataset.isnull().sum()
```

```python
dataset.columns
```

```python
# Pre_Processing the Dataset:
# Convert Categorical value into numerical value:
dataset = pd.get_dummies(dataset,drop_first=True)
dataset = dataset.astype(int)
dataset.head()
```

```python
# Split input output columns:
independent = dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
```

```python
dependent = dataset[['charges']]
```

```python
# Split train and test:
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
# Standardization:
#In simple linear regression's Coefficients are affected by feature scale. so, i am going to do StandardScaler.
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
# import Linear Regression model
from sklearn.linear_model import LinearRegression
# name the model:
regressor = LinearRegression()
# train the model:
regressor = regressor.fit(x_train,y_train)
regressor
# here model trained. 
```

```python
regressor.coef_
```

```python
regressor.intercept_
```

```python
# training predicted output:
y_train_predict = regressor.predict(x_train)
y_train_predict
```

```python
# train r2_score:
from sklearn.metrics import r2_score
train_r_score = r2_score(y_train,y_train_predict)
train_r_score
```

```python
# cheak the model prediction:
y_test_predict = regressor.predict(x_test)
y_test_predict
```

```python
#test r2_score:
test_r_score = r2_score(y_test,y_test_predict)
test_r_score
```

```python
print("Training R²:", train_r_score)
print("Test R²:", test_r_score)
```

```python
# Model Deployement:
# Save the model name:
filename = 'Linear_regression_Assign.sav'
```

```python
import pickle
pickle.dump(regressor,open(filename,'wb'))
```

```python
loaded_model = pickle.load(open('Linear_regression_Assign.sav','rb'))
loaded_model
```

```python
# Final checking:
new_data = pd.DataFrame({
    'age': [40],
    'bmi': [30.2],
    'children': [2],
    'sex_male': [1],
    'smoker_yes': [1]
})

new_data_scaled = sc.transform(new_data)
result = loaded_model.predict(new_data_scaled)
result
```
