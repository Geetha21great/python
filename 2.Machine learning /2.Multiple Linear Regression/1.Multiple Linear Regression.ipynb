# Multi Regression Assignment #

```python
# import libraries:
import pandas as pd
```

```python
#Reading dataset:
dataset=pd.read_csv('Startups.csv')
dataset.head()
```

```python
# one column is categorical so we are going to convert this as number.it a nominal data so we go for one-hot encoding.
dataset=pd.get_dummies(dataset,drop_first=True)
```

```python
dataset.head()
```

```python
#split input and output:
dataset.columns
```

```python
independent=dataset[['R&D Spend', 'Administration', 'Marketing Spend',
       'State_Florida', 'State_New York']]
independent.head()
```

```python
dependent=dataset[['Profit']]
dependent.head()
```

```python
#split train and test data:
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
# model selection
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor
```

```python
# Training or learning:
regressor.fit(x_train,y_train)
```

```python
# find weight:
weight = regressor.coef_
weight
```

```python
# find bias
bias = regressor.intercept_
bias
```

```python
# evaluate the learning:
from sklearn.metrics import r2_score
y_train_pred = regressor.predict(x_train)
train_r2 = r2_score(y_train, y_train_pred)
train_r2
```

```python
# model is ready for testing:
test_y_pred = regressor.predict(x_test)
test_y_pred
```

```python
# Evaluate the perforace comared to learning:
y_train_pred = regressor.predict(x_train)
train_r2 = r2_score(y_train, y_train_pred)
train_r2
```

```python
# r2 value is near to 1 so it is a good model.
```

```python
# store the modelname in one file:
filename = 'final_model_multi_linear_regression.sav'
```

```python
# dump the model in named file:
import pickle
pickle.dump(regressor,open(filename,'wb'))
```
