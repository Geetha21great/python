```python
# Import libraries:
import numpy as np 
import pandas as pd # It handling tables,reading file.
import matplotlib.pyplot as plt
```

```python
# read .csv file:
data = pd.read_csv('Salary_Data.csv')
data
```

```python
# Split input and output:
independent = data[['YearsExperience']]
dependent = data[['Salary']]
```

```python
# split dataset for training and testing:
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(independent,dependent,test_size=0.30,random_state=0)
```

```python
# Learning process:
from sklearn.linear_model import LinearRegression# import linearRegression
regressor = LinearRegression()#it assign in the name of regressor
#train the model:
regressor.fit(x_train,y_train)# here learning or training is happen
```

```python
# linear formula is y = wx+b
# w(weight/slope)
# bias
```

```python
weight=regressor.coef_
weight
```

```python
bias=regressor.intercept_
bias
```

```python
# Test or predict:
y_pred=regressor.predict(x_test)
y_pred
```

```python
# Evaluation metrics:
# r2 score only 0 to 1... r2 is nearly 1 means it is a good model:
from sklearn.metrics import r2_score
r_score = r2_score(y_test,y_pred)# r2 = actual value/predicted value
r_score
```

```python
# 0.9 nearly one so, it's a very good model
```

```python
# Save the model:
filename = 'finalized_model_linear.sav'# assigning the file name
```

```python
import pickle
pickle.dump(regressor,open(filename,'wb'))
# dump(write) the model in that particular named file.
```

```python
# load the model for final check:
loaded_model=pickle.load(open('finalized_model_linear.sav','rb'))
loaded_model
```
